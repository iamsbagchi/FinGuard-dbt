```python
from databricks.sdk import WorkspaceClient

# 1. Initialize client
w = WorkspaceClient()
scope_name = "finguard-scope"

# 2. Define your multiple secrets
secrets_to_add = {
    "bootstrap_servers": "",
    "api_key": "",
    "api_secret": "",
    "topic": "",
    "gmail_api_key": ""
}

# 3. Loop and add each key-value pair to the scope
for key, value in secrets_to_add.items():
    w.secrets.put_secret(scope=scope_name, key=key, string_value=value)
    print(f"Successfully added key: {key}")
```
