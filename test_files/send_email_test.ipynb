## Screipt to check for email test using gmail smtp server and configured google app password (from secret scope)

```python
body = """
<html>
<body>

<h2>Test Email from Databricks</h2>

<p>This is a test email sent from Databricks notebook.</p>

<p>If you receive this, the email configuration is working correctly.</p>

<br>

<p>
Thanks<br>
<b>FinGuard Support Team</b>
</p>

</body>
</html>
"""
subject = "Test Email from Databricks"
```

```python
gmail_api_key = dbutils.secrets.get(
            scope="finguard-scope",
            key="gmail_api_key"
        )
print(gmail_api_key)
```

```python
import smtplib
from email.mime.text import MIMEText
from email.mime.multipart import MIMEMultipart

# Gmail account <edit in your account details>
EMAIL = ""
APP_PASSWORD = gmail_api_key

# Receiver <edit in your account details>
to_email = ""

# Create message
msg = MIMEMultipart()
msg["From"] = EMAIL
msg["To"] = to_email
msg["Subject"] = subject

msg.attach(MIMEText(body, "html"))

try:
    with smtplib.SMTP("smtp.gmail.com", 587) as server:
        server.starttls()
        server.login(EMAIL, APP_PASSWORD)
        server.send_message(msg)

    print("✅ Email sent successfully!")

except Exception as e:
    print("❌ Error sending email:", e)
```
